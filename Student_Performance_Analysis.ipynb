```python
import glob
import pandas as pd

files = glob.glob("student_data/**/student*mat.csv", recursive=True)

print("Found files:", files)

df = pd.read_csv(files[0], sep=";")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

```python
# Display dataset information
print("Dataset Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
df.describe()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Average First Grade:", df["G1"].mean())
print("Average Second Grade:", df["G2"].mean())
print("Average Final Grade:", df["G3"].mean())
```

```python
import matplotlib.pyplot as plt

grades = ["G1", "G2", "G3"]
averages = [df["G1"].mean(), df["G2"].mean(), df["G3"].mean()]

plt.bar(grades, averages)

plt.title("Average Student Grades")
plt.xlabel("Examinations")
plt.ylabel("Average Marks")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(data=df, x="sex", y="G3")

plt.title("Gender-wise Student Performance")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")

plt.show()
```

```python
sns.barplot(data=df, x="studytime", y="G3")

plt.title("Study Time vs Student Performance")
plt.xlabel("Weekly Study Time")
plt.ylabel("Average Final Grade")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

correlation = df[["studytime", "failures", "absences", "G1", "G2", "G3"]].corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Between Student Factors and Grades")
plt.show()
```

```python
print("Highest Final Grade:", df["G3"].max())
print("Lowest Final Grade:", df["G3"].min())

print("Average Final Grade:", round(df["G3"].mean(), 2))
```
